# **2) CARGA y PREPROCESAMIENTO DE CIFAR-10**
En esta seccion se convierten las imágenes a tensores y normalizarlas.
* PyTorch: usar transforms.Normalize() por canal.
* TensorFlow: dividir por 255 y opcionalmente usar Standardization.
* Separar conjuntos de entrenamiento y prueba, asegurando balance de clases.

## **PYTOCH carga y preprocesamiento**

Para la carga y preprocesamiento de implemento lo siguiente:
## **1. Parametros generales**
Se define el tamano del batch (cantidad de imagenes procesadas por paso) y los valores de media y desviacion estandar del dataset CIFAR-10.(estos fueron encontrados en internet)  
Estos valores se usan para normalizar las imagenes y ayudar al modelo a entrenar de forma estable.

## **2. Transformaciones**
Estas operaciones preparan y mejoran los datos antes del entrenamiento.

### Transformaciones de entrenamiento(aplicando la tecnica de DATA AUMENTATION)
- **RandomHorizontalFlip**: voltea las imagenes horizontalmente con probabilidad 0.5.  
  Ayuda a que el modelo sea mas robusto.

- **RandomCrop**: recorta la imagen de forma aleatoria manteniendo el tamano original (32x32) con un borde extra (padding 4).  
  Esto simula variaciones naturales en las imagenes.

- **ToTensor**: convierte la imagen a formato tensor (PyTorch).

- **Normalize**: normaliza cada canal usando la media y desviacion estandar de CIFAR-10.  
  Esto mejora la estabilidad del entrenamiento y acelera la convergencia.

### Transformaciones de prueba
Solo se usan **ToTensor** y **Normalize**, ya que no se debe modificar la imagen durante la evaluacion.

## **3. Carga del dataset**
Se descargan o cargan las imagenes de CIFAR-10 aplicando las transformaciones:

- `train_dataset_pt`: conjunto de entrenamiento.
- `test_dataset_pt`: conjunto de prueba.

## **4. DataLoaders**
Los DataLoaders organizan los datos en batches y permiten iterarlos eficientemente:

- **train_loader**: carga los datos de entrenamiento en batches y los mezcla (shuffle=True) para mejorar el aprendizaje(asi no aprende de la secuencia o orden de los datos).
- **test_loader**: carga los datos de prueba sin mezclar (shuffle=False).


```python
# ============================
# 1. Parametros generales
# ============================
BATCH_SIZE = 128

CIFAR_MEAN = (0.4914, 0.4822, 0.4465)
CIFAR_STD  = (0.2023, 0.1994, 0.2010)

# ============================
# 2. Transformaciones
# ============================
# Aplicación de Data Augmentation
transform_train = torchvision.transforms.Compose([
    torchvision.transforms.RandomHorizontalFlip(p=0.5),
    torchvision.transforms.RandomCrop(32, padding=4),
    torchvision.transforms.ToTensor(),
    torchvision.transforms.Normalize(CIFAR_MEAN, CIFAR_STD),
])

transform_test = torchvision.transforms.Compose([
    torchvision.transforms.ToTensor(),
    torchvision.transforms.Normalize(CIFAR_MEAN, CIFAR_STD),
])

# ============================
# 3. Carga del dataset
# ============================
train_dataset_pt = torchvision.datasets.CIFAR10(
    root="./data", train=True, download=True, transform=transform_train
)

test_dataset_pt = torchvision.datasets.CIFAR10(
    root="./data", train=False, download=True, transform=transform_test
)

# ============================
# 4. DataLoaders
# ============================
train_loader = DataLoader(train_dataset_pt, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
test_loader  = DataLoader(test_dataset_pt,  batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

print("PyTorch - Imagenes de entrenamiento:", len(train_dataset_pt))
print("PyTorch - Imagenes de prueba:", len(test_dataset_pt))
```

### **PYTORCH Balanceo de Datos**  
En esta seccion de visualiza si los datos estan balanceados.
El balanceo es importante porque evita que el modelo aprenda mas sobre unas clases que sobre otras.  
Si una clase tiene muchas mas imagenes, la red puede volverse dependiente de esa clase y fallar al predecir las menos representadas.  

```python
# 1. Contar etiquetas del dataset
train_labels = [label for _, label in train_dataset_pt]
label_counts = Counter(train_labels)

# 2. Crear DataFrame
df_plot = pd.DataFrame({
    "Clase": [train_dataset_pt.classes[i] for i in range(10)],
    "Cantidad": [label_counts[i] for i in range(10)]
})

# 3. Gráfico mejorado
fig = px.bar(
    df_plot,
    x="Clase",
    y="Cantidad",
    text="Cantidad",
    title="Distribucion de Imagenes por Clase — CIFAR-10 (Entrenamiento)",
    labels={"Clase": "Categoria", "Cantidad": "Numero de Imagenes"},
)

# ---- ESTILOS MEJORADOS ----
fig.update_traces(
    textposition="outside",
    marker=dict(line=dict(width=1.5, color="black"))
)

fig.update_layout(

    title_x=0.5,
    font=dict(size=18),
    bargap=0.2,
    plot_bgcolor="white",
    xaxis=dict(showgrid=False),
    yaxis=dict(
        gridcolor="lightgray",
        range=[0, max(df_plot["Cantidad"]) * 1.25]
    ),
)

fig.show()
```

### **PYTORCH Ejemplos por cada categoria**

```python
# Funcion para desnormalizar la imagen
def imshow(img):
    img = img.numpy().transpose((1, 2, 0))  # (C,H,W) -> (H,W,C)
    img = img * np.array(CIFAR_STD) + np.array(CIFAR_MEAN)  # desnormalizar
    img = np.clip(img, 0, 1)  # asegurar rango [0,1]
    return img

# Clases del dataset
classes = train_dataset_pt.classes
num_classes = len(classes)

# Crear un diccionario para guardar la primera imagen de cada clase
class_images = {}

# Iterar sobre el DataLoader hasta tener una imagen de cada clase
for images, labels in train_loader:
    for img, label in zip(images, labels):
        if label.item() not in class_images:
            class_images[label.item()] = img
        if len(class_images) == num_classes:
            break
    if len(class_images) == num_classes:
        break

# Configurar figura
fig, axes = plt.subplots(1, num_classes, figsize=(num_classes*2, 2.5))

# Mostrar las imagenes
for i, label in enumerate(range(num_classes)):
    img = imshow(class_images[label])
    axes[i].imshow(img)
    axes[i].set_title(classes[label], fontsize=10)
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```

## **TENSORFLOW carga y preprocesamiento**

```python
# ============================
# 1. Carga del dataset
# ============================
(x_train_tf, y_train_tf), (x_test_tf, y_test_tf) = tf.keras.datasets.cifar10.load_data()


# Copia de imágenes reales (sin normalizar)
x_train_raw = x_train_tf.copy()
x_test_raw  = x_test_tf.copy()

# ============================
# 2 Normalización de las imágenes (Dividir por 255)
# ============================
x_train_tf = x_train_tf.astype('float32') / 255.0
x_test_tf = x_test_tf.astype('float32') / 255.0

# ============================
# 3 Preparación de etiquetas
# ============================
y_train_tf = y_train_tf.reshape(-1)
y_test_tf = y_test_tf.reshape(-1)

# ============================
# 4. NORMALIZACIÓN CON MEDIA Y STD DE CIFAR-10
# ============================
CIFAR_MEAN_TF = np.array([0.4914, 0.4822, 0.4465])
CIFAR_STD_TF = np.array([0.2023, 0.1994, 0.2010])

x_train_tf = (x_train_tf - CIFAR_MEAN_TF) / CIFAR_STD_TF
x_test_tf = (x_test_tf - CIFAR_MEAN_TF) / CIFAR_STD_TF

# ============================
# 5. AUMENTO DE DATOS (DATA AUGMENTATION)
# ============================
def augment_data(images, labels):
    # Flip horizontal
    images = tf.image.random_flip_left_right(images)

    # Padding correcto para imagen individual (rank 3)
    images = tf.pad(images, [[4, 4], [4, 4], [0, 0]], mode='REFLECT')

    # Recorte aleatorio de vuelta a 32x32
    images = tf.image.random_crop(images, size=[32, 32, 3])

    return images, labels

# ============================
# 6. CREACIÓN DE DATASETS CON AUMENTO DE DATOS
# ============================
BATCH_SIZE = 128

train_dataset = tf.data.Dataset.from_tensor_slices((x_train_tf, y_train_tf))
train_dataset = train_dataset.map(augment_data, num_parallel_calls=tf.data.AUTOTUNE)
train_dataset = train_dataset.shuffle(10000)
train_dataset = train_dataset.batch(BATCH_SIZE)
train_dataset = train_dataset.prefetch(tf.data.AUTOTUNE)

test_dataset = tf.data.Dataset.from_tensor_slices((x_test_tf, y_test_tf))
test_dataset = test_dataset.batch(BATCH_SIZE)
test_dataset = test_dataset.prefetch(tf.data.AUTOTUNE)
```

# **3) IMPLEMENTACION DE UN DNN (Multilayer Perceptron)**

## **PYTORCH DNN**





### Modelo DNN: Tecnicas Implementadas

El modelo `ModeloDNN` combina varias **tecnicas de Deep Learning** que permiten entrenar de manera eficiente y mejorar la generalizacion en tareas de clasificacion de imagenes.  

- **Flatten**  
    - Convierte las imagenes de entrada de tamaño `(3, 32, 32)` en un vector unidimensional de 3072 elementos.  
    - Permite pasar la informacion a las **capas totalmente conectadas**.

- **Capas Fully Connected (Lineales)**  
    - Aprenden combinaciones lineales de las caracteristicas de entrada.  
    - Extraen representaciones jerarquicas de la informacion.

- **He/Kaiming Initialization**  
    - Inicializacion de pesos especialmente diseñada para activaciones ReLU.  
    - Asegura que las activaciones no se saturen al inicio y que el gradiente fluya correctamente.

- **Batch Normalization**   
    - Normaliza las activaciones para que tengan media cero y desviacion estandar uno.  
    - Estabiliza y acelera el entrenamiento, y reduce la sensibilidad a la inicializacion de pesos.

- **Residual Connections**    
    - Inspiradas en ResNets, suman la entrada transformada de cada bloque a la salida de la capa lineal.  
    - Facilitan el flujo de gradientes y evitan el desvanecimiento del gradiente en redes profundas.

- **ReLU Activation**  
    - Funcion de activacion no lineal que introduce no linealidad en la red.  
    - Evita saturaciones y permite un aprendizaje mas rapido.

- **Dropout**  
    - Apaga aleatoriamente un porcentaje de neuronas (por ejemplo, 30%) durante el entrenamiento.  
    - Previene sobreajuste y obliga a la red a no depender de neuronas especificas.


```python

class ModeloDNN(nn.Module):
    def __init__(self, dropout_prob=0.3):
        super().__init__()
        self.flatten = nn.Flatten()

        # Capa 1: 3072 → 1024
        self.fc1 = nn.Linear(3*32*32, 1024)
        self.bn1 = nn.BatchNorm1d(1024)

        # Capa 2: 1024 → 512
        self.fc2 = nn.Linear(1024, 512)
        self.bn2 = nn.BatchNorm1d(512)
        self.res2 = nn.Linear(1024, 512)


        # Capa 3: 512 → 512
        self.fc3 = nn.Linear(512, 512)
        self.bn3 = nn.BatchNorm1d(512)
        self.res3 = nn.Identity()


        # Capa 4: 512 → 256
        self.fc4 = nn.Linear(512, 256)
        self.bn4 = nn.BatchNorm1d(256)
        self.res4 = nn.Linear(512, 256)


        # Capa 5: 256 → 128
        self.fc5 = nn.Linear(256, 128)
        self.bn5 = nn.BatchNorm1d(128)
        self.res5 = nn.Linear(256, 128)


        # Capa 6: 128 → 64
        self.fc6 = nn.Linear(128, 64)
        self.bn6 = nn.BatchNorm1d(64)
        self.res6 = nn.Linear(128, 64)


        # Capa de salida
        self.fc_out = nn.Linear(64, 10)

        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(dropout_prob)

        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Linear):
                nn.init.kaiming_normal_(m.weight)
                nn.init.zeros_(m.bias)

    def forward(self, x):
        x = self.flatten(x)

        # Flujo secuencial simple
        out = self.relu(self.bn1(self.fc1(x)))
        out = self.dropout(out)

        out = self.relu(self.bn2(self.fc2(out)+self.res2(out)))
        out = self.dropout(out)

        out = self.relu(self.bn3(self.fc3(out)+self.res3(out)))
        out = self.dropout(out)

        out = self.relu(self.bn4(self.fc4(out)+self.res4(out)))
        out = self.dropout(out)

        out = self.relu(self.bn5(self.fc5(out)+self.res5(out)))
        out = self.dropout(out)

        out = self.relu(self.bn6(self.fc6(out)+self.res6(out)))
        out = self.dropout(out)

        out = self.fc_out(out)
        return out


# Definir dispositivo (se reutiliza para todos los modelos PyTorch)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Dispositivo utilizado: {device}")

model_dnn_pt = ModeloDNN(dropout_prob=0.3).to(device)

print(model_dnn_pt)
```

### **ENTRENAMIENTO**

```python
# PARAMETROS DE ENTRENAMIENTO
# ============================
# 1. parametros generales
# ============================
EPOCHS = 90
LEARNING_RATE = 0.0001
EARLY_STOP_PATIENCE = 5  # número de épocas sin mejora antes de detener
EARLY_STOP_DELTA = 1e-4   # mínima mejora para resetear paciencia

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model_dnn_pt.parameters(), lr=LEARNING_RATE)

perdidas_entrenamiento, perdidas_prueba = [], []
precision_entrenamiento, precision_prueba = [], []

mejor_perdida = float('inf')
epocas_sin_mejora = 0

start_time_dnn_pt = time.time()

for epoch in range(EPOCHS):
    # ============================
    # Entrenamiento
    # ============================
    model_dnn_pt.train()
    for imagenes, etiquetas in train_loader:
        imagenes, etiquetas = imagenes.to(device, non_blocking=True), etiquetas.to(device, non_blocking=True)

        optimizer.zero_grad()
        salidas = model_dnn_pt(imagenes)
        perdida = criterion(salidas, etiquetas)

        perdida.backward()
        optimizer.step()

    # ============================
    # Evaluacion (train + test) - Cada época para mejor comparación
    # ============================
    model_dnn_pt.eval()

    # ---- Train ----
    acumulado_perdida, correctos, total = 0.0, 0, 0
    with torch.no_grad():
        for imagenes, etiquetas in train_loader:
            imagenes, etiquetas = imagenes.to(device, non_blocking=True), etiquetas.to(device, non_blocking=True)
            salidas = model_dnn_pt(imagenes)
            perdida = criterion(salidas, etiquetas)

            acumulado_perdida += perdida.item() * imagenes.size(0)
            _, prediccion = salidas.max(1)
            total += etiquetas.size(0)
            correctos += prediccion.eq(etiquetas).sum().item()

    perdida_train = acumulado_perdida / total
    precision_train = correctos / total
    perdidas_entrenamiento.append(perdida_train)
    precision_entrenamiento.append(precision_train)

    # ---- Test ----
    acumulado_perdida, correctos, total = 0.0, 0, 0
    with torch.no_grad():
        for imagenes, etiquetas in test_loader:
            imagenes, etiquetas = imagenes.to(device, non_blocking=True), etiquetas.to(device, non_blocking=True)
            salidas = model_dnn_pt(imagenes)
            perdida = criterion(salidas, etiquetas)

            acumulado_perdida += perdida.item() * imagenes.size(0)
            _, prediccion = salidas.max(1)
            total += etiquetas.size(0)
            correctos += prediccion.eq(etiquetas).sum().item()

    perdida_test = acumulado_perdida / total
    precision_test = correctos / total
    perdidas_prueba.append(perdida_test)
    precision_prueba.append(precision_test)

    # Imprimir cada 10 épocas para no saturar la salida
    if (epoch + 1) % 10 == 0 or epoch == 0:
        print(f"Epoch [{epoch+1}/{EPOCHS}]  "
              f"Train Loss: {perdida_train:.4f}, Train Acc: {precision_train:.4f}  "
              f"Test Loss: {perdida_test:.4f}, Test Acc: {precision_test:.4f}")

    # ============================
    # Early Stopping
    # ============================
    if perdida_test + EARLY_STOP_DELTA < mejor_perdida:
        mejor_perdida = perdida_test
        epocas_sin_mejora = 0
        torch.save(model_dnn_pt.state_dict(), "best_model_dnn_pt.pt")
    else:
        epocas_sin_mejora += 1
        if epocas_sin_mejora >= EARLY_STOP_PATIENCE:
            print(f"Early stopping: no mejora en {EARLY_STOP_PATIENCE} epocas.")
            break

training_time_dnn_pt = time.time() - start_time_dnn_pt

print(f"\nTiempo de entrenamiento DNN PyTorch: {training_time_dnn_pt:.2f} segundos")
print(f"Mejor precisión de prueba: {max(precision_prueba):.4f}")

# ============================
# Graficos
# ============================
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(range(1, len(perdidas_entrenamiento)+1), perdidas_entrenamiento, label='Loss Train')
plt.plot(range(1, len(perdidas_prueba)+1), perdidas_prueba, label='Loss Test')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Evolucion de la Perdida')
plt.legend()

plt.subplot(1,2,2)
plt.plot(range(1, len(precision_entrenamiento)+1), precision_entrenamiento, label='Acc Train')
plt.plot(range(1, len(precision_prueba)+1), precision_prueba, label='Acc Test')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Evolucion de la Precision')
plt.legend()

plt.tight_layout()
plt.show()
```

## **TENSORFLOW DNN**


### Modelo DNN: Arquitectura Similar a PyTorch

- **Batch Normalization**: Normaliza las activaciones para estabilizar el entrenamiento.

- **ReLU Activation**: Función de activación no lineal.

- **Dropout**: Previene sobreajuste apagando aleatoriamente neuronas durante el entrenamiento.


```python
# ============================
# Construcción del Modelo DNN
# ============================
model_dnn_tf = tf.keras.Sequential([
    # Flatten
    layers.Flatten(input_shape=(32, 32, 3)),
    
    # Capa 1: 3072 → 1024
    layers.Dense(1024, kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.Dropout(0.3),
    
    # Capa 2: 1024 → 512
    layers.Dense(512, kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.Dropout(0.3),
    
    # Capa 3: 512 → 512
    layers.Dense(512, kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.Dropout(0.3),
    
    # Capa 4: 512 → 256
    layers.Dense(256, kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.Dropout(0.3),
    
    # Capa 5: 256 → 128
    layers.Dense(128, kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.Dropout(0.3),
    
    # Capa 6: 128 → 64
    layers.Dense(64, kernel_initializer='he_normal'),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.Dropout(0.3),
    
    # Capa de salida: 64 → 10
    layers.Dense(10, activation='linear')  # Sin activación, se usa en la loss
])

# Compilar el modelo
model_dnn_tf.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy']
)

# Mostrar resumen del modelo
model_dnn_tf.summary()
```

# **4) IMPLEMENTACION DE UNA CNN (Red Neuronal Convolucional)**


## **PYTORCH CNN**


### Modelo CNN: Arquitectura Convolucional

El modelo CNN implementa las siguientes capas:

- **Capa Convolucional 1**: 32 filtros de 3x3 con padding, seguida de BatchNorm, ReLU y MaxPooling 2x2 (32x32 → 16x16)
- **Capa Convolucional 2**: 64 filtros de 3x3 con padding, seguida de BatchNorm, ReLU y MaxPooling 2x2 (16x16 → 8x8)
- **Capa Convolucional 3**: 128 filtros de 3x3 con padding, seguida de BatchNorm, ReLU y MaxPooling 2x2 (8x8 → 4x4)
- **Flatten**: Convierte el tensor multidimensional en un vector unidimensional (128 * 4 * 4 = 2048)
- **Capas Fully Connected**: 2048 → 512 → 256 → 128 → 10 (clasificación final)
- **Conexión Residual**: De la primera capa oculta (512) a la tercera capa oculta (128)
- **Dropout**: Para prevenir sobreajuste


```python
class ModeloCNN(nn.Module):
    def __init__(self, dropout_prob=0.3):
        super().__init__()
        
        # Primera capa convolucional
        self.conv1 = nn.Conv2d(3, 32, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(32)
        self.pool1 = nn.MaxPool2d(2, 2)  # 32x32 -> 16x16
        
        # Segunda capa convolucional
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(64)
        self.pool2 = nn.MaxPool2d(2, 2)  # 16x16 -> 8x8
        
        # Tercera capa convolucional
        self.conv3 = nn.Conv2d(64, 128, kernel_size=3, padding=1)
        self.bn3_conv = nn.BatchNorm2d(128)
        self.pool3 = nn.MaxPool2d(2, 2)  # 8x8 -> 4x4
        
        # Flatten
        self.flatten = nn.Flatten()
        
        # Capas fully connected
        # Después de 3 pooling de 2x2: 32x32 -> 16x16 -> 8x8 -> 4x4
        # 128 canales * 4 * 4 = 2048
        self.fc1 = nn.Linear(128 * 4 * 4, 512)  # Primera capa oculta
        self.bn3 = nn.BatchNorm1d(512)
        self.fc2 = nn.Linear(512, 256)  # Segunda capa oculta
        self.bn4 = nn.BatchNorm1d(256)
        self.fc3 = nn.Linear(256, 128)  # Tercera capa oculta
        self.bn5 = nn.BatchNorm1d(128)
        
        # Conexión residual: proyección de fc1 (512) a fc3 (128)
        self.fc_residual = nn.Linear(512, 128)
        
        self.fc_out = nn.Linear(128, 10)
        
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(dropout_prob)
        
        self._init_weights()
    
    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, (nn.Conv2d, nn.Linear)):
                nn.init.kaiming_normal_(m.weight)
                if m.bias is not None:
                    nn.init.zeros_(m.bias)
    
    def forward(self, x):
        # Primera capa convolucional
        x = self.conv1(x)
        x = self.bn1(x)
        x = self.relu(x)
        x = self.pool1(x)
        
        # Segunda capa convolucional
        x = self.conv2(x)
        x = self.bn2(x)
        x = self.relu(x)
        x = self.pool2(x)
        
        # Tercera capa convolucional
        x = self.conv3(x)
        x = self.bn3_conv(x)
        x = self.relu(x)
        x = self.pool3(x)
        
        # Flatten
        x = self.flatten(x)
        
        # Primera capa fully connected (guardar para conexión residual)
        fc1_out = self.relu(self.bn3(self.fc1(x)))
        x = self.dropout(fc1_out)
        
        # Segunda capa fully connected
        x = self.relu(self.bn4(self.fc2(x)))
        x = self.dropout(x)
        
        # Tercera capa fully connected con conexión residual
        fc3_out = self.relu(self.bn5(self.fc3(x)))
        # Conexión residual: sumar la proyección de fc1 a fc3
        residual = self.fc_residual(fc1_out)
        x = fc3_out + residual  # Conexión residual
        x = self.dropout(x)
        
        # Capa de salida
        x = self.fc_out(x)
        
        return x

# Crear modelo (usar el mismo device definido anteriormente)
model_cnn_pt = ModeloCNN(dropout_prob=0.3).to(device)

print(model_cnn_pt)

```

### **ENTRENAMIENTO**


```python
# PARAMETROS DE ENTRENAMIENTO
# ============================
EPOCHS_CNN = 90
LEARNING_RATE_CNN = 0.0001
EARLY_STOP_PATIENCE = 5
EARLY_STOP_DELTA = 1e-4

criterion_cnn = nn.CrossEntropyLoss()
optimizer_cnn = optim.Adam(model_cnn_pt.parameters(), lr=LEARNING_RATE_CNN)

perdidas_entrenamiento_cnn, perdidas_prueba_cnn = [], []
precision_entrenamiento_cnn, precision_prueba_cnn = [], []

mejor_perdida_cnn = float('inf')
epocas_sin_mejora_cnn = 0

start_time_cnn_pt = time.time()

for epoch in range(EPOCHS_CNN):
    # ============================
    # Entrenamiento
    # ============================
    model_cnn_pt.train()
    for imagenes, etiquetas in train_loader:
        imagenes, etiquetas = imagenes.to(device, non_blocking=True), etiquetas.to(device, non_blocking=True)

        optimizer_cnn.zero_grad()
        salidas = model_cnn_pt(imagenes)
        perdida = criterion_cnn(salidas, etiquetas)

        perdida.backward()
        optimizer_cnn.step()

    # ============================
    # Evaluacion (train + test) - Cada época para mejor comparación
    # ============================
    model_cnn_pt.eval()

    # ---- Train ----
    acumulado_perdida, correctos, total = 0.0, 0, 0
    with torch.no_grad():
        for imagenes, etiquetas in train_loader:
            imagenes, etiquetas = imagenes.to(device, non_blocking=True), etiquetas.to(device, non_blocking=True)
            salidas = model_cnn_pt(imagenes)
            perdida = criterion_cnn(salidas, etiquetas)

            acumulado_perdida += perdida.item() * imagenes.size(0)
            _, prediccion = salidas.max(1)
            total += etiquetas.size(0)
            correctos += prediccion.eq(etiquetas).sum().item()

    perdida_train = acumulado_perdida / total
    precision_train = correctos / total
    perdidas_entrenamiento_cnn.append(perdida_train)
    precision_entrenamiento_cnn.append(precision_train)

    # ---- Test ----
    acumulado_perdida, correctos, total = 0.0, 0, 0
    with torch.no_grad():
        for imagenes, etiquetas in test_loader:
            imagenes, etiquetas = imagenes.to(device, non_blocking=True), etiquetas.to(device, non_blocking=True)
            salidas = model_cnn_pt(imagenes)
            perdida = criterion_cnn(salidas, etiquetas)

            acumulado_perdida += perdida.item() * imagenes.size(0)
            _, prediccion = salidas.max(1)
            total += etiquetas.size(0)
            correctos += prediccion.eq(etiquetas).sum().item()

    perdida_test = acumulado_perdida / total
    precision_test = correctos / total
    perdidas_prueba_cnn.append(perdida_test)
    precision_prueba_cnn.append(precision_test)

    # Imprimir cada 10 épocas para no saturar la salida
    if (epoch + 1) % 10 == 0 or epoch == 0:
        print(f"Epoch [{epoch+1}/{EPOCHS_CNN}]  "
              f"Train Loss: {perdida_train:.4f}, Train Acc: {precision_train:.4f}  "
              f"Test Loss: {perdida_test:.4f}, Test Acc: {precision_test:.4f}")

    # ============================
    # Early Stopping
    # ============================
    if perdida_test + EARLY_STOP_DELTA < mejor_perdida_cnn:
        mejor_perdida_cnn = perdida_test
        epocas_sin_mejora_cnn = 0
        torch.save(model_cnn_pt.state_dict(), "best_model_cnn.pt")
    else:
        epocas_sin_mejora_cnn += 1
        if epocas_sin_mejora_cnn >= EARLY_STOP_PATIENCE:
            print(f"Early stopping: no mejora en {EARLY_STOP_PATIENCE} epocas.")
            break

training_time_cnn_pt = time.time() - start_time_cnn_pt

print(f"\nTiempo de entrenamiento CNN PyTorch: {training_time_cnn_pt:.2f} segundos")
print(f"Mejor precisión de prueba: {max(precision_prueba_cnn):.4f}")

# ============================
# Graficos
# ============================
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(range(1, len(perdidas_entrenamiento_cnn)+1), perdidas_entrenamiento_cnn, label='Loss Train')
plt.plot(range(1, len(perdidas_prueba_cnn)+1), perdidas_prueba_cnn, label='Loss Test')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Evolucion de la Perdida - CNN PyTorch')
plt.legend()

plt.subplot(1,2,2)
plt.plot(range(1, len(precision_entrenamiento_cnn)+1), precision_entrenamiento_cnn, label='Acc Train')
plt.plot(range(1, len(precision_prueba_cnn)+1), precision_prueba_cnn, label='Acc Test')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Evolucion de la Precision - CNN PyTorch')
plt.legend()

plt.tight_layout()
plt.show()
```

## **TENSORFLOW CNN**


### Modelo CNN: Arquitectura Similar a PyTorch

El modelo CNN de TensorFlow/Keras implementa la misma arquitectura que PyTorch:

- **Capa Convolucional 1**: 32 filtros de 3x3 con padding, BatchNorm, ReLU y MaxPooling 2x2 (32x32 → 16x16)
- **Capa Convolucional 2**: 64 filtros de 3x3 con padding, BatchNorm, ReLU y MaxPooling 2x2 (16x16 → 8x8)
- **Capa Convolucional 3**: 128 filtros de 3x3 con padding, BatchNorm, ReLU y MaxPooling 2x2 (8x8 → 4x4)
- **Flatten**: Convierte el tensor en un vector unidimensional (128 * 4 * 4 = 2048)
- **Capas Dense**: 2048 → 512 → 256 → 128 → 10
- **Conexión Residual**: De la primera capa oculta (512) a la tercera capa oculta (128)
- **Dropout**: Para prevenir sobreajuste

**Nota**: Se usa la API funcional de Keras para implementar la conexión residual.


```python
# ============================
# Construcción del Modelo CNN con API Funcional (para conexión residual)
# ============================
# Entrada
inputs = layers.Input(shape=(32, 32, 3))

# Primera capa convolucional
x = layers.Conv2D(32, (3, 3), padding='same', kernel_initializer='he_normal')(inputs)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)
x = layers.MaxPooling2D((2, 2))(x)  # 32x32 -> 16x16

# Segunda capa convolucional
x = layers.Conv2D(64, (3, 3), padding='same', kernel_initializer='he_normal')(x)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)
x = layers.MaxPooling2D((2, 2))(x)  # 16x16 -> 8x8

# Tercera capa convolucional
x = layers.Conv2D(128, (3, 3), padding='same', kernel_initializer='he_normal')(x)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)
x = layers.MaxPooling2D((2, 2))(x)  # 8x8 -> 4x4

# Flatten
x = layers.Flatten()(x)

# Primera capa fully connected (guardar para conexión residual)
fc1 = layers.Dense(512, kernel_initializer='he_normal')(x)
fc1 = layers.BatchNormalization()(fc1)
fc1_out = layers.ReLU()(fc1)
x = layers.Dropout(0.3)(fc1_out)

# Segunda capa fully connected
fc2 = layers.Dense(256, kernel_initializer='he_normal')(x)
fc2 = layers.BatchNormalization()(fc2)
fc2 = layers.ReLU()(fc2)
x = layers.Dropout(0.3)(fc2)

# Tercera capa fully connected con conexión residual
fc3 = layers.Dense(128, kernel_initializer='he_normal')(x)
fc3 = layers.BatchNormalization()(fc3)
fc3_out = layers.ReLU()(fc3)

# Conexión residual: proyección de fc1 (512) a fc3 (128)
fc1_residual = layers.Dense(128, kernel_initializer='he_normal')(fc1_out)
# Sumar la conexión residual
fc3_residual = layers.Add()([fc3_out, fc1_residual])
x = layers.Dropout(0.3)(fc3_residual)

# Capa de salida
outputs = layers.Dense(10, activation='linear')(x)  # Sin activación, se usa en la loss

# Crear el modelo
model_cnn_tf = models.Model(inputs=inputs, outputs=outputs)

# Compilar el modelo
model_cnn_tf.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy']
)

# Mostrar resumen del modelo
model_cnn_tf.summary()
```

### **ENTRENAMIENTO**


```python
# ============================
# Parámetros de Entrenamiento
# ============================
EPOCHS_CNN_TF = 90
LEARNING_RATE_CNN_TF = 0.0001

# Callbacks
early_stopping_cnn = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=5,
    min_delta=1e-4,
    restore_best_weights=True
)

# ============================
# Entrenamiento
# ============================
start_time_cnn_tf = time.time()

history_cnn_tf = model_cnn_tf.fit(
    train_dataset,
    epochs=EPOCHS_CNN_TF,
    validation_data=test_dataset,
    callbacks=[early_stopping_cnn],
    verbose=1
)

training_time_cnn_tf = time.time() - start_time_cnn_tf

# ============================
# Guardar métricas
# ============================
perdidas_entrenamiento_tf_cnn = history_cnn_tf.history['loss']
perdidas_prueba_tf_cnn = history_cnn_tf.history['val_loss']
precision_entrenamiento_tf_cnn = history_cnn_tf.history['accuracy']
precision_prueba_tf_cnn = history_cnn_tf.history['val_accuracy']

print(f"\nTiempo de entrenamiento CNN TensorFlow: {training_time_cnn_tf:.2f} segundos")
print(f"Mejor precisión de prueba: {max(precision_prueba_tf_cnn):.4f}")

# ============================
# Gráficos
# ============================
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(perdidas_entrenamiento_tf_cnn, label='Loss Train')
plt.plot(perdidas_prueba_tf_cnn, label='Loss Test')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Evolución de la Pérdida - CNN TensorFlow')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(precision_entrenamiento_tf_cnn, label='Acc Train')
plt.plot(precision_prueba_tf_cnn, label='Acc Test')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Evolución de la Precisión - CNN TensorFlow')
plt.legend()

plt.tight_layout()
plt.show()
```

## **Visualización de Filtros Aprendidos (Opcional - Nivel Avanzado)**

Visualización de los filtros aprendidos en la primera capa convolucional de la CNN de PyTorch.


```python
# ============================
# Visualización de Filtros de la Primera Capa Convolucional
# ============================
# Obtener los pesos de la primera capa convolucional
with torch.no_grad():
    # Cargar el mejor modelo si existe
    try:
        model_cnn_pt.load_state_dict(torch.load("best_model_cnn.pt"))
        print("Modelo cargado desde best_model_cnn.pt")
    except:
        print("Usando modelo actual (no se encontró best_model_cnn.pt)")

    # Obtener los filtros de la primera capa convolucional
    filters = model_cnn_pt.conv1.weight.data.cpu().numpy()
    
    # Normalizar los filtros para visualización
    filters = (filters - filters.min()) / (filters.max() - filters.min())

# Visualizar los primeros 32 filtros (o menos si hay menos)
num_filters = min(32, filters.shape[0])
n_cols = 8
n_rows = (num_filters + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 1.5, n_rows * 1.5))
axes = axes.flatten() if num_filters > 1 else [axes]

for i in range(num_filters):
    # Los filtros tienen forma (out_channels, in_channels, height, width)
    # Para visualizar, tomamos la media de los 3 canales RGB o mostramos cada canal
    filter_img = filters[i].transpose(1, 2, 0)  # (3, 3, 3) -> (3, 3, 3)
    # Promediar los 3 canales para visualización en escala de grises
    filter_img_gray = filter_img.mean(axis=2)
    
    axes[i].imshow(filter_img_gray, cmap='viridis')
    axes[i].set_title(f'Filtro {i+1}', fontsize=8)
    axes[i].axis('off')

# Ocultar ejes extra
for i in range(num_filters, len(axes)):
    axes[i].axis('off')

plt.suptitle('Filtros Aprendidos - Primera Capa Convolucional (CNN PyTorch)', fontsize=14, y=0.995)
plt.tight_layout()
plt.show()

# Visualización alternativa: mostrar los 3 canales RGB por separado para algunos filtros
print("\nVisualización de filtros RGB (primeros 4 filtros):")
fig, axes = plt.subplots(4, 3, figsize=(9, 12))
for i in range(min(4, filters.shape[0])):
    for channel in range(3):
        axes[i, channel].imshow(filters[i, channel], cmap='viridis')
        axes[i, channel].set_title(f'Filtro {i+1}, Canal {channel}', fontsize=8)
        axes[i, channel].axis('off')
plt.suptitle('Filtros RGB - Primera Capa Convolucional', fontsize=14)
plt.tight_layout()
plt.show()
```
