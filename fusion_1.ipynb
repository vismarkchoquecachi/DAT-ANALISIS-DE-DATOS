# **FUSION DE DATASETS**


El siguiente código realiza una **fusión interna** (`INNER JOIN`) entre los dos DataFrames (`datos_estudiantes` y `data_notas`) utilizando los identificadores únicos validados previamente.
## **Componentes de la Fusión**

| Componente | Rol en el DataFrame | Descripción |
| :--- | :--- | :--- |
| **`datos_estudiantes`** | **Izquierda** | Contiene la información demográfica, ubicación y el identificador **`dip`**. |
| **`data_notas`** | **Derecha** | Contiene las notas del examen y el identificador **`C.I.`**. |
| **`left_on="dip"`** | **Clave Izquierda** | La columna **`dip`** de la tabla izquierda se usará para la coincidencia. |
| **`right_on="C.I."`** | **Clave Derecha** | La columna **`C.I.`** de la tabla derecha se usará para la coincidencia. |
| **`how="inner"`** | **Tipo de Unión** | **Fusión Interna (`INNER JOIN`)**. Solo se incluyen las filas donde **`dip` es idéntico a `C.I.`**. |

```python
# Conjuntos únicos
set_notas = set(data_notas["C.I."])
set_dips  = set(datos_estudiantes["dip"])

# Intersección
coinciden_unicos = set_notas & set_dips

print("Cantidad de identificadores únicos que coinciden:", len(coinciden_unicos))
```

Se tiene 342 identificadores unicos "C.I." 

```python
fusion = datos_estudiantes.merge(
    data_notas,
    left_on="dip",
    right_on="C.I.",
    how="inner"
)
print("------------------------------------------------")
print("Total de filas fusionadas:", len(fusion))

print("------------------------------------------------")
print(fusion.info())
```


## **4. Desempeño de los postulantes por carrera**
 **1. Filtros interactivos**
Se incorporan **widgets interactivos** que permiten al usuario seleccionar:

- Una o varias **ciudades** de procedencia.
- Uno o varios **géneros**.

 **2. Cálculo de indicadores**
Para cada carrera, se calculan:

- **Total de postulantes**
- **Cantidad de aprobados** (Estado = 1)
- **Cantidad de reprobados** (Estado = 0)
- **Porcentaje de aprobación** (% Aprobación)


**Esto permite:**
- Permite comparar rápidamente **el rendimiento de los postulantes entre carreras**.
- Identifica **carreras con mayor o menor tasa de aprobación**.
- Facilita el análisis por **ciudad y género** gracias a los filtros interactivos.



```python
!pip install ipywidgets -q
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- UNA SOLA VEZ: pre-calculamos todo y copiamos solo lo necesario ---
df = fusion[["C.I.", "Carrera", "Estado", "Ciudad", "Género"]].copy()
df["Género"] = df["Género"].astype("category")
df["Ciudad"] = df["Ciudad"].astype("category")
df["Carrera"] = df["Carrera"].astype("category")
df["Estado"] = df["Estado"].astype("int8")  # 0 o 1

# Pre-calculamos el resumen completo una sola vez
resumen_total = (
    df.groupby(["Carrera", "Ciudad", "Género", "Estado"], observed=True)["C.I."]
    .count()
    .reset_index(name="Cantidad")
    .astype({"Cantidad": "int32"})
)

# Opciones disponibles
ciudades = list(df["Ciudad"].cat.categories)
generos = list(df["Género"].cat.categories)

# Creamos listas de checkboxes para cada filtro
ciudad_checkboxes = []
for ciudad in ciudades:
    cb = widgets.Checkbox(
        value=True,
        description=str(ciudad),
        indent=False,
        layout=widgets.Layout(width="220px")
    )
    cb._valor = ciudad  # guardamos el valor real de la categoría
    ciudad_checkboxes.append(cb)

genero_checkboxes = []
for genero in generos:
    cb = widgets.Checkbox(
        value=True,
        description=str(genero),
        indent=False,
        layout=widgets.Layout(width="160px")
    )
    cb._valor = genero
    genero_checkboxes.append(cb)

# Contenedores visuales estilo "tarjeta"
ciudad_box = widgets.VBox([
    widgets.HTML("<b>Ciudad</b>"),
    widgets.GridBox(
        ciudad_checkboxes,
        layout=widgets.Layout(
            grid_template_columns="repeat(2, 230px)",
            grid_gap="4px 12px"
        )
    )
])

genero_box = widgets.VBox([
    widgets.HTML("<b>Género</b>"),
    widgets.VBox(genero_checkboxes)
])

filtros_contenido = widgets.HBox(
    [ciudad_box, genero_box],
    layout=widgets.Layout(justify_content="space-around", padding="10px 20px")
)

filtros_card = widgets.VBox(
    [
        widgets.HTML("""
        <div style='font-size:16px; font-weight:bold; margin-bottom:4px;'>
            Filtros de Postulantes
        </div>
        <div style='font-size:12px; color:#555;'>
            Marca o desmarca opciones para actualizar el gráfico en tiempo real.
        </div>
        """),
        filtros_contenido
    ],
    layout=widgets.Layout(
        border="1px solid #ccc",
        padding="10px 15px",
        margin="5px 0 10px 0",
        border_radius="8px",
        width="100%"
    )
)

# Acordeón plegable para no ocupar tanto espacio en la vista
filtros_accordion = widgets.Accordion(children=[filtros_card])
filtros_accordion.set_title(0, "Filtros de Postulantes")
# Inicia cerrado (None). Si quieres que inicie abierto, cambia a 0
filtros_accordion.selected_index = None

salida_mejorada = widgets.Output()


def _valores_seleccionados(checks):
    """Devuelve la lista de valores (_valor) seleccionados en un grupo de checkboxes."""
    return [cb._valor for cb in checks if cb.value]


def actualizar_rapido_mejorado(_=None):
    with salida_mejorada:
        clear_output(wait=True)

        ciudades_sel = _valores_seleccionados(ciudad_checkboxes)
        generos_sel = _valores_seleccionados(genero_checkboxes)

        if not ciudades_sel or not generos_sel:
            print("Selecciona al menos una ciudad y un género para ver el gráfico.")
            return

        # Filtro ultra-rápido sobre el resumen ya pre-agrupado
        mask = (
            resumen_total["Ciudad"].isin(ciudades_sel)
            & resumen_total["Género"].isin(generos_sel)
        )
        temp = resumen_total[mask]

        if temp.empty:
            print("No hay datos con los filtros seleccionados")
            return

        # Total por carrera con los filtros actuales
        total_por_carrera = temp.groupby("Carrera", observed=True)["Cantidad"].sum()

        # Pivot para tener Aprobados y Reprobados
        pivot = (
            temp.pivot_table(
                index="Carrera",
                columns="Estado",
                values="Cantidad",
                aggfunc="sum",
                fill_value=0,
                observed=False,
            )
            .rename(columns={0: "Reprobados", 1: "Aprobados"})
        )

        for col in ["Aprobados", "Reprobados"]:
            if col not in pivot.columns:
                pivot[col] = 0

        pivot = pivot[["Aprobados", "Reprobados"]].astype(int)
        pivot["Total"] = total_por_carrera
        pivot["Pct_Aprobados"] = (pivot["Aprobados"] / pivot["Total"] * 100).round(1)
        pivot["Pct_Reprobados"] = (pivot["Reprobados"] / pivot["Total"] * 100).round(1)

        # Formato largo para Plotly
        plot_df = pivot.reset_index().melt(
            id_vars="Carrera",
            value_vars=["Aprobados", "Reprobados"],
            var_name="Estado",
            value_name="Cantidad",
            
        )
        plot_df["Porcentaje"] = plot_df.apply(
            lambda row: pivot.loc[row["Carrera"], f"Pct_{row['Estado']}"], axis=1
        )

        fig = px.bar(
            plot_df,
            x="Carrera",
            y="Cantidad",
            color="Estado",
            text=plot_df["Porcentaje"].round(1).astype(str) + "%",
            title="Desempeño de Postulantes por Carrera (filtros aplicados)",
            barmode="stack",
            color_discrete_map={"Aprobados": "#69D3BE", "Reprobados": "#ED695D"},
            hover_data={"Total": pivot["Total"].reindex(plot_df["Carrera"]).values},
        )

        fig.update_traces(textposition="inside")
        fig.update_layout(
            height=650,
            width=1100,
            plot_bgcolor="rgba(245,245,245,1)",
            margin=dict(t=60, l=40, r=20, b=40),
        )
        fig.show()


# Conectamos los cambios de cualquier checkbox a la actualización del gráfico
def _on_checkbox_change(change):
    if change["name"] == "value":
        actualizar_rapido_mejorado()

for cb in ciudad_checkboxes + genero_checkboxes:
    cb.observe(_on_checkbox_change, names="value")

# Mostramos el panel completo (con filtros plegables)
ui_mejorado = widgets.VBox([
    filtros_accordion,
    salida_mejorada,
])

display(ui_mejorado)

# Primera ejecución
actualizar_rapido_mejorado()
```

## **5. Desempeño de Postulantes por Administración del Colegio**
SE OBSERVA a continuación cómo se distribuyen los **postulantes** y su **desempeño académico** (aprobados y reprobados) según el tipo de **administración del colegio**, considerando filtros por **carrera**, **género** y **tipo de administración**.

 **Filtros interactivos**
Se incorporan **widgets interactivos** que permiten al usuario seleccionar:

- Una o varias **carreras**.
- Uno o varios **géneros**.
- Una o varias **administraciones de colegio** (público, privado, etc.).

Esto permite explorar los datos de manera dinámica y ver cómo varía el desempeño según diferentes combinaciones.

**Cálculo de indicadores**
Para cada tipo de administración del colegio se calculan:

- **Total de postulantes**
- **Cantidad de aprobados**
- **Cantidad de reprobados**
- **Porcentaje de aprobación** (% Aprobación)

Estos indicadores permiten entender rápidamente la proporción de éxito de los postulantes en cada administración.

**Esto permite:**

- Comparar el **desempeño de los postulantes entre distintos tipos de administración**.
- Identificar qué administración tiene **mayor o menor porcentaje de aprobación**.
- Analizar cómo influyen la **carrera** y el **género** gracias a los filtros interactivos.


```python
import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- Copiar DataFrame ---
df = fusion.copy()
df["Género"] = df["Género"].astype(str)
df["Carrera"] = df["Carrera"].astype(str)
df["Administración del Colegio"] = df["Administración del Colegio"].astype(str)

# ======================================================
#     GENERACIÓN DE CHECKBOXES POR CATEGORÍA
# ======================================================

def generar_checkbox_group(titulo, valores, width="220px"):
    checks = []
    for v in sorted(valores):
        cb = widgets.Checkbox(
            value=True,
            description=str(v),
            indent=False,
            layout=widgets.Layout(width=width)
        )
        cb._valor = v
        checks.append(cb)

    caja = widgets.VBox([
        widgets.HTML(f"<b>{titulo}</b>"),
        widgets.VBox(checks)
    ])
    return checks, caja


# Crear checkboxes por categoría
checks_carrera, carrera_box = generar_checkbox_group(
    "Carrera", df["Carrera"].unique(), width="240px"
)

checks_genero, genero_box = generar_checkbox_group(
    "Género", df["Género"].unique(), width="160px"
)

checks_admin, admin_box = generar_checkbox_group(
    "Administración del Colegio", df["Administración del Colegio"].unique(), width="300px"
)

# Contenedor visual estilo tarjeta
filtros_card = widgets.VBox(
    [
        widgets.HTML("""
        <div style='font-size:16px; font-weight:bold; margin-bottom:4px;'>
            Filtros
        </div>
        <div style='font-size:12px; color:#555;'>
            Marca o desmarca opciones para actualizar el gráfico.
        </div>
        """),
        widgets.HBox([carrera_box, genero_box, admin_box],
                     layout=widgets.Layout(gap="30px", padding="10px"))
    ],
    layout=widgets.Layout(
        border="1px solid #ccc",
        border_radius="8px",
        padding="10px 15px",
        width="100%"
    )
)

# Acordeón para ocultar/mostrar filtros
filtros_accordion = widgets.Accordion(children=[filtros_card])
filtros_accordion.set_title(0, "Filtros")
filtros_accordion.selected_index = None  # inicia cerrado

# Contenedor de salida
salida = widgets.Output()


# ======================================================
#    FUNCIONES AUXILIARES
# ======================================================

def valores_seleccionados(checks):
    return [cb._valor for cb in checks if cb.value]


# ======================================================
#    FUNCIÓN DE ANÁLISIS (NO MODIFICADA)
# ======================================================
def analizar_admin_aprobacion(df, carreras, generos, administraciones):
    df_filtrado = df[
        df["Carrera"].isin(carreras) &
        df["Género"].isin(generos) &
        df["Administración del Colegio"].isin(administraciones)
    ]
    
    total = df_filtrado.groupby("Administración del Colegio")["C.I."].count().reset_index()
    total.columns = ["Administración del Colegio", "Total"]
    
    aprobados = df_filtrado[df_filtrado["Estado"] == 1].groupby("Administración del Colegio")["C.I."].count().reset_index()
    aprobados.columns = ["Administración del Colegio", "Aprobados"]
    
    reprobados = df_filtrado[df_filtrado["Estado"] == 0].groupby("Administración del Colegio")["C.I."].count().reset_index()
    reprobados.columns = ["Administración del Colegio", "Reprobados"]
    
    resumen = total.merge(aprobados, on="Administración del Colegio", how="left").merge(reprobados, on="Administración del Colegio", how="left")
    resumen[["Aprobados", "Reprobados"]] = resumen[["Aprobados", "Reprobados"]].fillna(0).astype(int)
    
    resumen["Pct_Aprobados"] = (resumen["Aprobados"] / resumen["Total"] * 100).round(1)
    resumen["Pct_Reprobados"] = (resumen["Reprobados"] / resumen["Total"] * 100).round(1)
    resumen[["Pct_Aprobados","Pct_Reprobados"]] = resumen[["Pct_Aprobados","Pct_Reprobados"]].fillna(0)

    resumen_long = resumen.melt(
        id_vars=["Administración del Colegio", "Total", "Pct_Aprobados", "Pct_Reprobados"],
        value_vars=["Aprobados", "Reprobados"],
        var_name="Estado",
        value_name="Cantidad"
    )

    resumen_long["Pct"] = resumen_long.apply(
        lambda r: r["Pct_Aprobados"] if r["Estado"] == "Aprobados" else r["Pct_Reprobados"],
        axis=1
    )

    resumen_long["Administración del Colegio"] = pd.Categorical(
        resumen_long["Administración del Colegio"],
        categories=sorted(df["Administración del Colegio"].unique()),
        ordered=True
    )

    return resumen_long


# ======================================================
#    FUNCIÓN PARA ACTUALIZAR GRÁFICO
# ======================================================
def actualizar(_=None):
    with salida:
        clear_output(wait=True)

        carreras = valores_seleccionados(checks_carrera)
        generos = valores_seleccionados(checks_genero)
        admins = valores_seleccionados(checks_admin)

        if not carreras or not generos or not admins:
            print("Selecciona al menos un valor en cada categoría.")
            return

        resumen_long = analizar_admin_aprobacion(df, carreras, generos, admins)

        if resumen_long.empty:
            print("No hay datos disponibles con esos filtros.")
            return

        fig = px.bar(
            resumen_long,
            x="Administración del Colegio",
            y="Cantidad",
            color="Estado",
            text="Pct",
            title="Desempeño de Postulantes por Administración del Colegio",
            barmode="stack",
            color_discrete_map={"Aprobados": "#4DECCC", "Reprobados": "#E66969"},
            hover_data=["Total", "Pct", "Estado"]
        )

        fig.update_traces(
            texttemplate="%{text:.1f}%",
            textposition="inside",
            hovertemplate=(
                "<b>%{x}</b><br>"
                "Estado: %{customdata[2]}<br>"
                "Cantidad: %{y}<br>"
                "Total: %{customdata[0]}<br>"
                "Porcentaje del estado: %{customdata[1]}%<extra></extra>"
            )
        )

        fig.update_layout(
            height=650,
            width=1100,
            plot_bgcolor="rgba(245,245,245,1)",
            title_x=0.5
        )
        fig.show()


# ======================================================
#    CONECTAR CHECKBOXES A ACTUALIZACIÓN
# ======================================================
for cb in checks_carrera + checks_genero + checks_admin:
    cb.observe(lambda change: actualizar(), names="value")


# ======================================================
#    MOSTRAR UI COMPLETA
# ======================================================
display(widgets.VBox([filtros_accordion, salida]))

# Primera ejecución
actualizar()
```

## **6. Analisis de difultad por materia**
Este análisis muestra qué materias son más difíciles para los estudiantes según su carrera, gestión y género. Se calculan estadísticas como promedio, mediana, nota mínima y máxima, y porcentaje de reprobados. Los resultados se presentan mediante box plots y barras de promedio, permitiendo identificar de manera visual las materias con mayor dificultad y bajo rendimiento.

```python
import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display, clear_output

# ================================================
# 1) Columnas de materias
# ================================================
columnas_materias = [c for c in fusion.columns if c.startswith("Nota")]

# ================================================
# 2) Función de análisis
# ================================================
def analizar_dificultad_por_carrera(df, carrera, gestion, genero):
    df_car = df[(df["Carrera"] == carrera) & (df["Gestión"] == gestion)].copy()

    if genero != "Todos":
        df_car = df_car[df_car["Género"] == genero]

    resultados = []

    for col in columnas_materias:

        notas_validas = df_car[col][df_car[col] != -1]

        if notas_validas.empty:
            continue

        materia = col.replace("Nota", "")
        total = len(notas_validas)
        aprob = (notas_validas >= 51).sum()
        reprob = (notas_validas < 51).sum()

        resultados.append({
            "Materia": materia,
            "Promedio": notas_validas.mean(),
            "Mediana": notas_validas.median(),
            "Nota mínima": notas_validas.min(),
            "Nota máxima": notas_validas.max(),
            "Total estudiantes": total,
            "Aprobados": aprob,
            "Reprobados": reprob,
            "% Reprobados": (reprob / total) * 100
        })

    return pd.DataFrame(resultados)


# ================================================
# 3) Widgets
# ================================================
carreras = sorted(fusion["Carrera"].unique())
gestiones = sorted(fusion["Gestión"].unique())
generos = ["Todos"] + sorted(fusion["Género"].dropna().unique())

selector_carrera = widgets.Dropdown(
    options=carreras,
    description="Carrera:",
    layout=widgets.Layout(width="250px")
)

selector_gestion = widgets.Dropdown(
    options=gestiones,
    description="Gestión:",
    layout=widgets.Layout(width="150px")
)

selector_genero = widgets.Dropdown(
    options=generos,
    description="Género:",
    layout=widgets.Layout(width="150px")
)

salida = widgets.Output()


# ================================================
# 4) Función para actualizar gráficos
# ================================================
def actualizar_graficos(change):
    with salida:
        clear_output()

        carrera = selector_carrera.value
        gestion = selector_gestion.value
        genero = selector_genero.value

        df_res = analizar_dificultad_por_carrera(fusion, carrera, gestion, genero)

        if df_res.empty:
            print("No hay datos válidos para esta configuración.")
            return

        df_car = fusion[(fusion["Carrera"] == carrera) & (fusion["Gestión"] == gestion)].copy()
        if genero != "Todos":
            df_car = df_car[df_car["Género"] == genero]

        df_long = df_car.melt(
            id_vars=["Carrera"],
            value_vars=columnas_materias,
            var_name="Materia",
            value_name="Nota"
        )

        df_long = df_long[df_long["Nota"] != -1]
        df_long["Materia"] = df_long["Materia"].str.replace("Nota", "")

        # ================================
        # 1) BOX PLOT (categorical colors)
        # ================================
        fig_box = px.box(
            df_long,
            x="Materia",
            y="Nota",
            color="Materia",      # ← cada materia un color
            title=f"Distribución de notas – {carrera} ({gestion}) – {genero}",
            points="all"
        )
        fig_box.update_layout(yaxis_title="Nota", xaxis_title="Materia")
        fig_box.show()

        # ================================
        # 2) PROMEDIO (gradiente azul)
        # ================================
        fig_prom = px.bar(
            df_res.sort_values("Promedio"),
            x="Materia",
            y="Promedio",
            color="Promedio",               # ← degradado azul
            color_continuous_scale="Blues", # ← escala azul original
            title=f"Promedio de notas – {carrera} ({gestion}) – {genero}",
            text="Promedio"
        )
        fig_prom.update_traces(texttemplate='%{text:.1f}', textposition='outside')
        fig_prom.update_layout(yaxis_title="Promedio", xaxis_title="Materia")
        fig_prom.show()


# ================================================
# 5) Conectar eventos
# ================================================
selector_carrera.observe(actualizar_graficos, names="value")
selector_gestion.observe(actualizar_graficos, names="value")
selector_genero.observe(actualizar_graficos, names="value")

# ================================================
# 6) Mostrar interfaz
# ================================================
display(widgets.HBox([selector_carrera, selector_gestion, selector_genero]), salida)

# Ejecutar primera vez
actualizar_graficos(None)
```

Además, el gráfico es completamente interactivo: permite hacer zoom, mostrar porcentajes y ver detalles al pasar el cursor.
```python
import plotly.express as px
from IPython.display import display

# === PREPARACIÓN RÁPIDA Y BONITA ===
df = fusion.copy()

# Nombres lindos y claros
df["Género"] = df["Género"].str.replace("MASCULINO", "Hombres").replace("FEMENINO", "Mujeres")
df["Tipo"] = df["Administración del Colegio"].replace({"FISCAL": "Colegio Fiscal", "PRIVADA": "Colegio Privado"})
df["Resultado"] = df["Estado"].map({1: "Aprobó", 0: "No Aprobó"})

# Contar para que los cuadrados tengan el tamaño correcto
conteo = df.groupby(["Tipo", "Género", "Carrera", "Resultado"]).size().reset_index(name="Estudiantes")

# === EL ÁRBOL MÁGICO ===
fig = px.treemap(
    conteo,
    path=["Tipo", "Género", "Carrera", "Resultado"],  # ¡Jerarquía perfecta!
    values="Estudiantes",
    color="Resultado",
    color_discrete_map={"Aprobó": "#27AE60", "No Aprobó": "#E74C3C"},
    title="El Árbol de la Vida Universitaria UMSA<br>"
          "<sup>→ Pasa el mouse sobre cualquier cuadrado • Haz clic para hacer zoom • Cada cuadrado = cantidad de estudiantes</sup>",
    height=750,
    template="plotly_white"
)

# ¡Hacerlo AÚN MÁS bonito!
fig.update_traces(
    textinfo="label+value+percent parent",
    textfont_size=16,
    hovertemplate="<b>%{label}</b><br>Estudiantes: %{value}<br>Porcentaje: %{percentParent:.1%}<extra></extra>"
)

fig.update_layout(
    margin=dict(t=100, l=0, r=0, b=0),
    title_x=0.5,
    font=dict(family="Arial Black", size=18, color="#2c3e50")
)

fig.show()
```
