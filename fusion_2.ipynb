# **FUSION DE DATASETS**


El siguiente código realiza una **fusión interna** (`INNER JOIN`) entre los dos DataFrames (`datos_estudiantes` y `data_notas`) utilizando los identificadores únicos validados previamente.

## **Componentes de fusion**

Antes de la fusión, es esencial entender los valores únicos en las columnas clave de ambos *DataFrames* para prever problemas de formato, codificación o discrepancias que podrían afectar cualquier análisis bivariado posterior.

| Variable | DataFrame Base (`fusion_1` - Datos Pre-Fusionados) | DataFrame Encuesta (`datos_examen` - Encuesta) | Estado de los Datos |
| :--- | :--- | :--- | :--- |
| **Género** | **`fusion_1["Género"]`** | **`datos_examen["Genero al que pertenece"]`** | Los formatos de texto deben coincidir para agrupar (ej. 'MASCULINO' vs 'Masculino'). |
| **Edad** | **`fusion_1["Edad"]`** | **`datos_examen["Escriba su edad en años cumplidos"]`** | Se debe asegurar que los valores de `datos_examen` sean numéricos e iguales para comparar con la `Edad` calculada en `fusion_1`. |
| **Nivel de Aprendizaje** | **`fusion_1["nivel de aprendizaje"]`** | **`datos_examen["El nivel de aprendizaje que ha logrado fue..."]`** | Ambos están en una escala 1-5, lo que permite la comparación directa. |

```python
'''## **Perfiles de Estudiantes (Modelo basado en transición de estados / tipo Markov)**

Para clasificar el rendimiento y comportamiento académico se usó un sistema de perfiles basado en reglas lógicas que representan posibles *estados* del estudiante.  
Cada estado puede interpretarse como un nivel o fase dentro del proceso de aprendizaje (similar a un modelo de transición tipo Markov):

- **reprobado**  
  Estudiantes con notas muy bajas (< 40) pero con nivel de aprendizaje reportado alto (≥ 4).  
  Representa un estado donde el estudiante *cree que aprende*, pero el rendimiento no lo confirma.

- **Éxito Real**  
  Estudiantes con notas altas (> 70) y nivel de aprendizaje alto (≥ 4).  
  Es el estado ideal: aprendizaje y resultados coherentes.

- **En Riesgo**  
  Estudiantes con promedio < 50 independientemente del nivel de aprendizaje.  
  Estado crítico con alta probabilidad de reprobar.

- **Estándar**  
  Estudiantes que no cumplen ninguna de las condiciones anteriores.  
  Estado estable sin riesgo inmediato.'''
```



### **Gráfico: Distribución de Perfiles de Estudiantes**

El gráfico generado muestra cuántos estudiantes caen en cada estado.  
Esto permite analizar:

- proporción de riesgo real,  
- diferencia entre percepción de aprendizaje y el rendimiento,  
- y posibles transiciones entre estados (por ejemplo, “En Riesgo → Estándar → Éxito Real”).

Este análisis es útil para modelos educativos, seguimiento académico y simulaciones basadas en estados.


```python
cond = [
    (df['promedio_aprobacion'] < 40) & (df['nivel de aprendizaje'] >= 4),
    (df['promedio_aprobacion'] > 70) & (df['nivel de aprendizaje'] >= 4),
    (df['promedio_aprobacion'] < 50)
]
df['Perfil'] = np.select(cond, ['Autoengaño', 'Éxito Real', 'En Riesgo'], default='Estándar')

plt.figure()
ax = sns.countplot(data=df, x='Perfil', hue='Perfil',
                   palette=['#e84393', '#00cec9', '#fab1a0', '#74b9ff'],
                   order=['Autoengaño', 'Éxito Real', 'En Riesgo', 'Estándar'],
                   legend=False)

plt.title('PERFILES DE ESTUDIANTES EN CLASES VIRTUALES', fontweight='bold', pad=20)
plt.ylabel('Cantidad de Estudiantes')
plt.xlabel('')
for p in ax.patches:
    ax.text(p.get_x() + p.get_width()/2, p.get_height() + 1,
            f'{int(p.get_height())}', ha='center', fontsize=50, fontweight='bold')
plt.tight_layout()
plt.show()
```

Este gráfico compara la cantidad de materias aprobadas por los estudiantes según su acceso a tecnología en casa.  
Para ello se combinan dos variables de la encuesta:  
- Si el estudiante tiene computadora, laptop o tablet.  
- Si cuenta con acceso a internet (wifi o datos móviles).

El boxplot permite observar cómo varía el rendimiento académico entre los distintos niveles de acceso tecnológico.  
Es útil para identificar si disponer de computadora e internet está asociado con aprobar más materias.


```python
# =========================================================
# Gráfico Adicional 4: Materias Aprobadas vs Acceso a Tecnología
# =========================================================
ex = datos_examen.copy()
ex['Acceso_Tecnologia'] = ex['En su casa usted dispone de computadora personal, laptop o Tablet?'] + ' + ' + \
                          ex['Qué tipo de acceso a internet tiene desde su casa?'].apply(lambda x: 'Internet' if x in ['wifi','datos moviles'] else 'Sin Internet')

plt.figure(figsize=(16, 10))
sns.boxplot(data=ex, x='Acceso_Tecnologia', y='Cantidad de materias aprobadas hasta la fecha', palette='Set3')
plt.title('MATERIAS APROBADAS SEGÚN ACCESO A TECNOLOGÍA', pad=40)
plt.xlabel('Acceso a PC/Internet')
plt.ylabel('Materias Aprobadas')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
